```python
import numpy as np

A = np.array([[56.0, 0.0, 4.4, 68.0], [1.2, 104.0, 52.0, 8.0], [1.8, 135.0, 99.0, 0.9]])

print(A)
```

```python
cal = A.sum(axis = 0)
#axis = 0 means sum vertically, axis = 1 sums horizontally
print(cal)
```

```python
#reshape commands runs in O(1) time, so dont be shy to use it
#we can use it to make sure the matrix is the size we want it to be
percentage = 100 * A/cal.reshape(1, 4)
#this is an example of broadcasting, we divided the 3, 4 matrix by a 1, 4 matrix
#basically, what happens is, python auto expands the 1, 4 into a 3, 4, makes 2 more copies of it, and then each row is divided
print(percentage)
```
